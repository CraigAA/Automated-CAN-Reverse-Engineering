```python
import torch
from torch.utils.data import Dataset, DataLoader
import pandas as pd
import numpy as np

class CANClassificationDataset(Dataset):
    def __init__(self, bit_data_list, labels_list, start_indices_list, window_size=100):
        self.window_size = window_size
        self.bit_data_list = bit_data_list
        self.labels_list = labels_list
        
        # Flatten the start indices so PyTorch can just ask for item 'idx'
        self.samples = []
        for array_id, start_idx_array in enumerate(start_indices_list):
            for start_idx in start_idx_array:
                self.samples.append((array_id, start_idx))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        # 1. Look up which CAN ID array to slice, and where to start
        array_id, start_idx = self.samples[idx]
        
        # 2. Slice the 200 frames exactly when needed (0 RAM overhead)
        window_data = self.bit_data_list[array_id][start_idx : start_idx + self.window_size]
        base_labels = self.labels_list[array_id]
        
        # 3. Dynamic Masking calculated on-the-fly!
        # ptp (peak-to-peak) finds the difference between max and min. If 0, it's static.
        static_mask = np.ptp(window_data, axis=0) == 0 
        target_y = base_labels.copy()
        target_y[static_mask] = 0
        
        return torch.tensor(window_data, dtype=torch.float32), torch.tensor(target_y, dtype=torch.long)


def get_labeled_dataloaders(file_path, labels_dict, train_ids, val_ids, test_ids, window_size=100, window_stride=100, batch_size=256):
    print("Loading raw CSV...")
    df = pd.read_csv(file_path, sep='\s+', header=None, names=['timestamp', 'bus', 'raw_frame'])
    
    def hex_to_bits(h):
        return [int(b) for b in bin(int(h, 16))[2:].zfill(64)]

    def process_ids(id_list):
        bit_data_list, labels_list, start_indices_list = [], [], []
        
        for target_id in id_list:
            if target_id not in labels_dict: continue
            
            # Filter
            df_target = df[df['raw_frame'].str.contains(f"^{target_id}#")]
            if len(df_target) < window_size: continue
            
            # Extract
            payloads = df_target['raw_frame'].str.split('#').str[1]
            # Convert directly to int8 array
            bit_data = np.array(list(payloads.apply(hex_to_bits)), dtype=np.int8) 
            
            # Calculate valid starting blocks (0, 10, 20...)
            max_start = bit_data.shape[0] - window_size
            start_indices = np.arange(0, max_start + 1, window_stride)
            
            # Store the raw, lightweight data
            bit_data_list.append(bit_data)
            labels_list.append(np.array(labels_dict[target_id], dtype=np.int8))
            start_indices_list.append(start_indices)
            
        if not bit_data_list: return None
        return CANClassificationDataset(bit_data_list, labels_list, start_indices_list, window_size)

    print("Building Train Dataset...")
    train_ds = process_ids(train_ids)
    print("Building Validation Dataset...")
    val_ds = process_ids(val_ids)
    print("Building Test Dataset...")
    test_ds = process_ids(test_ids)
    
    train_dl = DataLoader(train_ds, batch_size=batch_size, shuffle=True) if train_ds else None
    val_dl = DataLoader(val_ds, batch_size=batch_size, shuffle=False) if val_ds else None
    test_dl = DataLoader(test_ds, batch_size=batch_size, shuffle=False) if test_ds else None
    
    return train_dl, val_dl, test_dl
```

```python
import torch
import torch.nn as nn

class TemporalBitTransformer(nn.Module):
    def __init__(self, d_model=32, nhead=2, num_layers=2, 
                 dim_feedforward=64, num_classes=5, dropout=0.2):
        super(TemporalBitTransformer, self).__init__()
        
        self.num_classes = num_classes

        # 1. Input Projection: Note the change here! 
        # We are no longer projecting 64 bits. We are projecting a SINGLE bit's 
        # value (0 or 1) at a single time step into the d_model space.
        self.input_projection = nn.Linear(1, d_model)
        
        # 2. The Transformer Encoder (Unchanged)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # 3. Final Classification Head: Note the change!
        # We no longer output 64 * num_classes. We just output num_classes 
        # for the single bit time-series currently being evaluated.
        self.fc = nn.Linear(d_model, num_classes)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # Initial shape: (Batch, TimeSteps, 64) -> e.g., (256, 50, 64)
        batch_size, seq_len, num_bits = x.size()

        # --- THE CRITICAL SHIFT: ISOLATING THE BITS ---
        
        # 1. Swap Time and Bits dimensions -> (Batch, 64, 50)
        x = x.permute(0, 2, 1)
        
        # 2. Fold the Bits into the Batch dimension -> (Batch * 64, 50)
        # Now we have 256 * 64 = 16,384 completely independent time-series.
        # The network shares weights across all 64 bits, forcing translation invariance!
        x = x.reshape(batch_size * num_bits, seq_len)
        
        # 3. Add a feature dimension for the Linear layer -> (Batch * 64, 50, 1)
        # Because nn.Linear expects (Batch, Seq_Len, Features)
        x = x.unsqueeze(-1)

        # --- STANDARD TRANSFORMER PROCESSING ---
        
        # Project: (Batch * 64, 50, d_model)
        x = self.input_projection(x)  
        
        # Self-Attention: (Batch * 64, 50, d_model)
        x = self.transformer_encoder(x) 

        # Sequence Pooling (Average the 50 time steps): (Batch * 64, d_model)
        x = x.mean(dim=1)              

        # Decode to predictions: (Batch * 64, num_classes)
        x = self.dropout(x)
        out = self.fc(x)               

        # --- RECONSTRUCT THE ORIGINAL SHAPE ---
        
        # Unfold the predictions back into their respective bit positions
        # Final Shape: (Batch, 64, num_classes)
        out = out.view(batch_size, num_bits, self.num_classes)

        return out

def Transformer_baseline(dropout=0.2, lr=0.001, d_model=128, nhead=4, 
                         num_layers=3, dim_feedforward=256, num_classes=5):
    
    # Swap out CANTransformerClassifier for the new model
    model = TemporalBitTransformer(
        d_model=d_model,
        nhead=nhead,
        num_layers=num_layers,
        dim_feedforward=dim_feedforward,
        num_classes=num_classes,
        dropout=dropout
    )
    
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    manual_weights = torch.tensor([0.5, 10.0, 3.0, 0.5, 2.5], dtype=torch.float).to(device)
    criterion = nn.CrossEntropyLoss(weight=manual_weights, label_smoothing=0.05)

    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
    return model, criterion, optimizer
```

```python
def universal_training_loop(model_initializer, train_dl, val_dl, epochs=50, 
                            device=None, verbose=True, **model_kwargs):
    from sklearn.metrics import f1_score
    
    if device is None:
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    if verbose:
        print(f"Using device: {device}")

    # Initialize whichever model you passed in (LSTM or Transformer)
    model, criterion, optimizer = model_initializer(**model_kwargs)
    model = model.to(device)

    train_losses, val_losses, train_accs, val_accs = [], [], [], []
    all_val_preds = []
    all_val_targets = []
    
    for epoch in range(epochs):
        # Training Phase
        model.train()
        train_loss, train_correct, train_total = 0.0, 0, 0

        for data, target in train_dl:
            data, target = data.to(device), target.to(device)
            optimizer.zero_grad()
            
            outputs = model(data)
            
            # Reshape for loss calculation
            outputs = outputs.view(-1, model_kwargs.get('num_classes', 5))
            target = target.view(-1)

            loss = criterion(outputs, target)
            loss.backward()
            optimizer.step()

            train_loss += loss.item()
            _, predicted = torch.max(outputs.data, 1)
            train_total += target.size(0)
            train_correct += (predicted == target).sum().item()

        avg_train_loss = train_loss / len(train_dl)
        train_acc = 100. * train_correct / train_total

        # Validation Phase
        model.eval()
        val_loss, val_correct, val_total = 0.0, 0, 0

        with torch.no_grad():
            for data, target in val_dl:
                data, target = data.to(device), target.to(device)
                
                outputs = model(data)
                outputs = outputs.view(-1, model_kwargs.get('num_classes', 5))
                target = target.view(-1)

                loss = criterion(outputs, target)
                val_loss += loss.item()
                
                _, predicted = torch.max(outputs.data, 1)
                val_total += target.size(0)
                val_correct += (predicted == target).sum().item()
                # Move to CPU, convert to numpy, flatten to 1D, and store
                all_val_preds.extend(predicted.cpu().numpy().flatten())
                all_val_targets.extend(target.cpu().numpy().flatten())
                
        avg_val_loss = val_loss / len(val_dl)
        val_acc = 100. * val_correct / val_total
        epoch_val_loss = val_loss / len(val_dl)

        # Calculate Macro F1-Score
        # average='macro' gives equal weight to all classes, preventing Class 0 
        # from dominating the score.
        val_macro_f1 = f1_score(all_val_targets, all_val_preds, average='macro', zero_division=0)
        
        train_losses.append(avg_train_loss)
        val_losses.append(avg_val_loss)
        train_accs.append(train_acc)
        val_accs.append(val_acc)

        # Print progress
        if verbose and (epoch + 1) % 1 == 0:
            print(f'Epoch [{epoch+1}/{epochs}], '
                  f'Train Loss: {avg_train_loss:.4f}',
                  f'Val Loss: {avg_val_loss:.4f}',
                  f'Val F1: {val_macro_f1:.4f}')
    if verbose:
        print(f'Final Results - Train Loss: {train_losses[-1]:.4f}, Train Acc: {train_accs[-1]:.2f}%, '
              f'Val Loss: {val_losses[-1]:.4f}, Val Acc: {val_accs[-1]:.2f}%')
    return model, train_losses, val_losses, train_accs, val_accs
```

```python
import torch
import numpy as np
from sklearn.metrics import accuracy_score, f1_score

def get_final_prediction(model, dataloader, num_bits=64, num_classes=5, device='cuda'):
    """
    Evaluates a single CAN ID across all temporal windows using Majority Voting.
    Returns the final 64-bit prediction and vote matrix.
    """
    model.eval()
    vote_matrix = torch.zeros((num_bits, num_classes), dtype=torch.int64).to(device)
    
    with torch.no_grad():
        for data, target in dataloader:
            data = data.to(device)
            outputs = model(data)
            
            # Handle both 3D and 2D model output shapes
            if len(outputs.shape) == 4:
                _, preds = torch.max(outputs[:, -1, :, :], dim=2) 
            else:
                _, preds = torch.max(outputs, dim=2)
            
            data_np = data.cpu().numpy()
            
            # Mask-Aware Voting: Find bits that were completely static in this 100-frame window
            is_static_mask = np.ptp(data_np, axis=1) == 0 
            
            # Tally votes only for bits that actively flipped
            for bit_idx in range(num_bits):
                active_preds = preds[~is_static_mask[:, bit_idx], bit_idx]
                
                if len(active_preds) > 0:
                    bit_votes = torch.bincount(active_preds, minlength=num_classes)
                    vote_matrix[bit_idx] += bit_votes
                
    # Calculate the final layout via argmax
    final_predicted_layout = np.zeros(num_bits, dtype=int)
    vote_matrix_np = vote_matrix.cpu().numpy()
    
    for bit_idx in range(num_bits):
        total_votes = np.sum(vote_matrix_np[bit_idx])
        if total_votes == 0:
            # If a bit NEVER flipped across the entire log, it defaults to Static/Spare (0)
            final_predicted_layout[bit_idx] = 0
        else:
            final_predicted_layout[bit_idx] = np.argmax(vote_matrix_np[bit_idx])
            
    return final_predicted_layout, vote_matrix_np
```

```python
from sklearn.metrics import accuracy_score, f1_score
import numpy as np
import warnings

def get_test_evaluation(final_predicted_layout, ground_truth_layout):
    """
    Calculates metrics, returning np.nan for architectures not present in the sequence.
    """
    pred_arr = np.array(final_predicted_layout)
    gt_arr = np.array(ground_truth_layout)
    
    # 1. Overall Accuracy
    accuracy = accuracy_score(gt_arr, pred_arr) * 100.0
    
    # 2. Extract Per-Class F1
    # We use np.nan for zero_division so missing classes return NaN instead of 0.0
    # Suppress the UndefinedMetricWarning since we are handling it intentionally
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        per_class_f1 = f1_score(gt_arr, pred_arr, labels=[1, 2, 3, 4], average=None, zero_division=np.nan)
        
    bit_f1, counter_f1, sensor_f1, checksum_f1 = per_class_f1
    
    # 3. Dynamic Macro F1 (ignoring NaN values)
    # np.nanmean calculates the average of the array, completely ignoring the np.nan values
    # In your example of [NaN, 1.0, 1.0, NaN], this correctly returns 1.0
    macro_f1 = np.nanmean(per_class_f1)
    
    # Edge case: If the sequence was entirely empty (all NaNs), nanmean returns NaN. 
    # You might want to default that to 0.0 or leave it as NaN depending on your preference.
    if np.isnan(macro_f1):
        macro_f1 = 0.0 
        
    return accuracy, macro_f1, bit_f1, counter_f1, sensor_f1, checksum_f1

# --- Example Print Formatting ---
# To print this out clearly in your terminal, you can format the NaN values like this:
def format_metric(val):
    return "N/A" if np.isnan(val) else f"{val:.2f}"

# Usage:
# print(f"Macro: {format_metric(macro_f1)} | Bit: {format_metric(bit_f1)} | Counter: {format_metric(counter_f1)} ...")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

def plot_final_prediction(predicted_layout, ground_truth, vote_matrix, can_id="Unknown"):
    """
    Generates plot comparing the Ground Truth to the Final Prediction, followed by confidence heatmaps for each class.
    """
    fig, ax = plt.subplots(figsize=(15, 6.5))
    ax.set_title(f"Architectural Decoding & Vote Confidence: ID {can_id}", fontsize=16, pad=25)

    # Class Mapping
    class_names = {
        1: '1: Bit Flag', 
        2: '2: Counter', 
        3: '3: Sensor', 
        4: '4: Checksum'
    }
    class_colors = {
        1: '#e74c3c',  # Red/Crimson
        2: '#2ecc71',  # Green
        3: '#3498db',  # Blue
        4: '#f39c12'   # Orange
    }

    byte_spacing = 9  # 8 bits + 1 unit gap for visual clarity
    row_height = 1
    
    # Calculate vote confidences (0.0 to 1.0) to use as alpha values for the heatmaps
    row_sums = vote_matrix.sum(axis=1, keepdims=True)
    row_sums[row_sums == 0] = 1 # Prevent division by zero for completely dead bits
    confidence_matrix = vote_matrix / row_sums

    y_cursor = 0

    # Draw Byte Labels at the very top
    for byte_index in range(8):
        x_center = byte_index * byte_spacing + 4
        ax.text(x_center, y_cursor + 1.2, f"Byte {byte_index}", ha="center", fontsize=10, fontweight='bold')

    # Ground Truth & Final Prediction
    layouts = [("Ground Truth", ground_truth), ("Final Prediction", predicted_layout)]

    for label, layout in layouts:
        ax.text(-1, y_cursor + 0.5, label, ha="right", va="center", fontsize=11, fontweight='bold')
        
        for bit_idx, class_id in enumerate(layout):
            byte_index = bit_idx // 8
            bit_inner = bit_idx % 8
            x = byte_index * byte_spacing + bit_inner

            color = class_colors.get(class_id, '#ffffff')
            rect = patches.Rectangle((x, y_cursor), 1, row_height,
                                     facecolor=color, edgecolor="black", linewidth=1)
            ax.add_patch(rect)
            
        y_cursor -= 1.2

    # Confidence Heatmaps 
    y_cursor -= 0.2 
    ax.text(-1, y_cursor + 0.5, "Vote Confidence", ha="right", va="center", fontsize=12, fontweight='bold', style='italic')
    y_cursor -= 1.2
    
    for class_idx in range(1,5):
        c_name = class_names[class_idx]
        c_color = class_colors[class_idx]
        
        # Label the row
        ax.text(-1, y_cursor + 0.5, c_name, ha="right", va="center", fontsize=10, fontweight='bold', color=c_color if class_idx != 0 else 'black')

        for bit_idx in range(64):
            byte_index = bit_idx // 8
            bit_inner = bit_idx % 8
            x = byte_index * byte_spacing + bit_inner
            
            conf_score = confidence_matrix[bit_idx, class_idx]

            if conf_score > 0:
                rect = patches.Rectangle((x, y_cursor), 1, row_height,
                                         facecolor=c_color,
                                         alpha=float(conf_score),
                                         edgecolor="none")
                ax.add_patch(rect)
            rect_border = patches.Rectangle((x, y_cursor), 1, row_height,
                                            facecolor="none",
                                            edgecolor="lightgray", linewidth=0.5)
            ax.add_patch(rect_border)

        y_cursor -= 1.2
    ax.set_xlim(-15, 8 * byte_spacing)
    ax.set_ylim(y_cursor + 0.5, 2.5)
    ax.axis("off")
    plt.tight_layout()
    plt.show()
```

```python
print("\n==================================================")
print("ARCHITECTURAL DECODING: ID 002 (Impreza Holdout)")
print("==================================================") 

predicted_layout, vote_matrix = get_final_prediction(
    model=tf_model,
    dataloader=test_dl,
    device='cuda'
)

ground_truth = Impreza_Ground_Labels['002']

accuracy, macro_f1, bit_f1, counter_f1, sensor_f1, checksum_f1 = get_test_evaluation(predicted_layout, ground_truth)
print('{:.2f}%, {:.2f} macro, {:.2f} bit, {:.2f} counter, {:.2f} sensor, {:.2f} checksum'.format(accuracy, macro_f1, bit_f1, counter_f1, sensor_f1, checksum_f1))

plot_final_prediction(
    predicted_layout=predicted_layout, 
    ground_truth=ground_truth, 
    vote_matrix=vote_matrix, 
    can_id="002 (Impreza)"
)
```
